```python
def Desc_stat(ds,var):
    mean=ds[var].mean()
    median=ds[var].median()
    mode=ds[var].mode()[0]
    minimum= ds[var].min()
    maximum=ds[var].max()
    range = ds[var].max() - ds[var].min()
    std_dev= ds[var].std()
    variance= ds[var].var()
    skew = ds[var].skew()
    kurtosis = ds[var].kurt()
    count = ds[var].count()
    Quart = [
        ds[var].quantile(0.25),
        ds[var].quantile(0.50),
        ds[var].quantile(0.75),
        ds[var].quantile(0.75) - ds[var].quantile(0.25)
    ]

    summary = {
    "Average": mean,
        "Median": median,
        "Mode": mode,
        "Minimum": minimum,
        "Maximum": maximum,
        "Range": range,
        "Std Dev": std_dev,
        "Variance": variance,
        "Skewness": skew,
        "Kurtosis": kurtosis,
        "25th Percentile": Quart[0],
        "50th Percentile": Quart[1],
        "75th Percentile": Quart[2],
        "Inter Quartile Range": Quart[3],
        "Count": count
}
    return summary 
```

```python
num_cols = ["math percentage", "reading score percentage", "writing score percentage"]
```

```python
df[num_cols] = df[num_cols].fillna(df[num_cols].mean())
```

```python
Q1_r = df["reading score percentage"].quantile(0.25)
Q3_r = df["reading score percentage"].quantile(0.75)
IQR_r = Q3_r - Q1_r
lower_r = Q1_r - 1.5 * IQR_r
upper_r = Q3_r + 1.5 * IQR_r
print("Reading lower limit:", lower_r)
print("Reading upper limit:", upper_r)
```

```python
Q1_w = df["writing score percentage"].quantile(0.25)
Q3_w = df["writing score percentage"].quantile(0.75)
IQR_w = Q3_w - Q1_w
lower_w = Q1_w - 1.5 * IQR_w
upper_w = Q3_w + 1.5 * IQR_w
print("Writing lower limit:", lower_w)
print("Writing upper limit:", upper_w)
```

```python
df[df["reading score percentage"] < lower_r]
```

```python
df[df["writing score percentage"] < lower_w]
```

```python
df = df.drop([59, 76, 211, 327, 596, 980])
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.model_selection import train_test_split

X = df.iloc[:, [0, 1, 2, 3, 5, 6, 7]].values
y = df.iloc[:, 4].values

ct = ColumnTransformer(transformers=[('encoder', OneHotEncoder(drop='first'), [0, 1, 2, 3, 6])], remainder='passthrough')
X = np.array(ct.fit_transform(X))

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)
print(X_train.shape)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

regressor = LinearRegression()
regressor.fit(X_train, y_train)
y_pred_lr = regressor.predict(X_test)

print("Linear Regression MSE:", mean_squared_error(y_test, y_pred_lr))
print("Linear Regression R2:", r2_score(y_test, y_pred_lr))
```

```python
from sklearn.preprocessing import PolynomialFeatures

poly_reg = PolynomialFeatures(degree=2)
X_poly_train = poly_reg.fit_transform(X_train)
X_poly_test = poly_reg.transform(X_test)

poly_model = LinearRegression()
poly_model.fit(X_poly_train, y_train)
y_pred_poly = poly_model.predict(X_poly_test)

print("Polynomial Regression MSE:", mean_squared_error(y_test, y_pred_poly))
print("Polynomial Regression R2:", r2_score(y_test, y_pred_poly))
```

```python
from sklearn.linear_model import Ridge

rdge = Ridge(alpha=1.0)
rdge.fit(X_train, y_train)
y_pred_ridge = rdge.predict(X_test)

print("Ridge Regression MSE:", mean_squared_error(y_test, y_pred_ridge))
print("Ridge Regression R2:", r2_score(y_test, y_pred_ridge))
```

```python
from sklearn.linear_model import Lasso

las = Lasso(alpha=0.001)
las.fit(X_train, y_train)
y_pred_lasso = las.predict(X_test)

print("Lasso Regression MSE:", mean_squared_error(y_test, y_pred_lasso))
print("Lasso Regression R2:", r2_score(y_test, y_pred_lasso))
```

```python
from sklearn.linear_model import ElasticNet

enet = ElasticNet(alpha=0.001, l1_ratio=0.5)
enet.fit(X_train, y_train)
y_pred_enet = enet.predict(X_test)

print("Elastic Net Regression MSE:", mean_squared_error(y_test, y_pred_enet))
print("Elastic Net Regression R2:", r2_score(y_test, y_pred_enet))
```

```python
threshold = 0.6
df["Result"] = np.where(df["math percentage"] >= threshold, 1, 0)

print("Number of Students Passed:", (df["Result"] == 1).sum())
print("Number of Students Failed:", (df["Result"] == 0).sum())
df.head()
```

```python
from sklearn.preprocessing import StandardScaler

y_class = df["Result"].values

X_train_c, X_test_c, y_train_c, y_test_c = train_test_split(X, y_class, test_size=0.2, random_state=1)

sc = StandardScaler()
X_train_c = sc.fit_transform(X_train_c)
X_test_c = sc.transform(X_test_c)
print(X_train_c.shape)
print(X_test_c.shape)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score, f1_score

lr = LogisticRegression(random_state=0)
lr.fit(X_train_c, y_train_c)
y_pred_lr_c = lr.predict(X_test_c)

cm = confusion_matrix(y_test_c, y_pred_lr_c)
print(cm)
print("Errors (FP + FN):", cm[0, 1] + cm[1, 0])
print("Accuracy:", accuracy_score(y_test_c, y_pred_lr_c))
print("Precision:", precision_score(y_test_c, y_pred_lr_c))
print("Recall:", recall_score(y_test_c, y_pred_lr_c))
print("F1-score:", f1_score(y_test_c, y_pred_lr_c))
```

```python
from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors=5, metric='minkowski', p=2)
knn.fit(X_train_c, y_train_c)
y_pred_knn = knn.predict(X_test_c)

cm = confusion_matrix(y_test_c, y_pred_knn)
print(cm)
print("Errors (FP + FN):", cm[0, 1] + cm[1, 0])
print("Accuracy:", accuracy_score(y_test_c, y_pred_knn))
print("Precision:", precision_score(y_test_c, y_pred_knn))
print("Recall:", recall_score(y_test_c, y_pred_knn))
print("F1-score:", f1_score(y_test_c, y_pred_knn))
```

```python
from sklearn.naive_bayes import GaussianNB

nb = GaussianNB()
nb.fit(X_train_c, y_train_c)
y_pred_nb = nb.predict(X_test_c)

cm = confusion_matrix(y_test_c, y_pred_nb)
print(cm)
print("Errors (FP + FN):", cm[0, 1] + cm[1, 0])
print("Accuracy:", accuracy_score(y_test_c, y_pred_nb))
print("Precision:", precision_score(y_test_c, y_pred_nb))
print("Recall:", recall_score(y_test_c, y_pred_nb))
print("F1-score:", f1_score(y_test_c, y_pred_nb))
```

```python
from sklearn.tree import DecisionTreeClassifier

dt = DecisionTreeClassifier(criterion='entropy', random_state=0)
dt.fit(X_train_c, y_train_c)
y_pred_dt = dt.predict(X_test_c)

cm = confusion_matrix(y_test_c, y_pred_dt)
print(cm)
print("Errors (FP + FN):", cm[0, 1] + cm[1, 0])
print("Accuracy:", accuracy_score(y_test_c, y_pred_dt))
print("Precision:", precision_score(y_test_c, y_pred_dt))
print("Recall:", recall_score(y_test_c, y_pred_dt))
print("F1-score:", f1_score(y_test_c, y_pred_dt))
```

```python
from sklearn.svm import SVC

svm = SVC(kernel='rbf', random_state=0)
svm.fit(X_train_c, y_train_c)
y_pred_svm = svm.predict(X_test_c)

cm = confusion_matrix(y_test_c, y_pred_svm)
print(cm)
print("Errors (FP + FN):", cm[0, 1] + cm[1, 0])
print("Accuracy:", accuracy_score(y_test_c, y_pred_svm))
print("Precision:", precision_score(y_test_c, y_pred_svm))
print("Recall:", recall_score(y_test_c, y_pred_svm))
print("F1-score:", f1_score(y_test_c, y_pred_svm))
```

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(n_estimators=100, criterion='entropy', random_state=0)
rf.fit(X_train_c, y_train_c)
y_pred_rf = rf.predict(X_test_c)

cm = confusion_matrix(y_test_c, y_pred_rf)
print(cm)
print("Errors (FP + FN):", cm[0, 1] + cm[1, 0])
print("Accuracy:", accuracy_score(y_test_c, y_pred_rf))
print("Precision:", precision_score(y_test_c, y_pred_rf))
print("Recall:", recall_score(y_test_c, y_pred_rf))
print("F1-score:", f1_score(y_test_c, y_pred_rf))
```

```python
models = ["Logistic Regression", "K-Nearest Neighbors", "Naive Bayes", "Decision Tree", "SVM", "Random Forest"]
predictions = [y_pred_lr_c, y_pred_knn, y_pred_nb, y_pred_dt, y_pred_svm, y_pred_rf]

results_c = []
for name, pred in zip(models, predictions):
    cm = confusion_matrix(y_test_c, pred)
    results_c.append({"Model": name,
                      "Confusion Matrix": cm.tolist(),
                      "Errors (FP + FN)": cm[0, 1] + cm[1, 0],
                      "Accuracy": accuracy_score(y_test_c, pred),
                      "Precision": precision_score(y_test_c, pred),
                      "Recall": recall_score(y_test_c, pred),
                      "F1-Score": f1_score(y_test_c, pred)})

results_c = pd.DataFrame(results_c)
print(results_c)

best_c = results_c.loc[results_c["F1-Score"].idxmax()]
print("\nBest-Fit Model:", best_c["Model"])
print("Accuracy:", best_c["Accuracy"])
print("F1-Score:", best_c["F1-Score"])
```
